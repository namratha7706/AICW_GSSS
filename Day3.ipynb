```python
# array operations
arr2=np.array([1,2,3])
arr3=np.array([4,5,6])
add_array = arr2+arr3
mul_array=arr2*arr3
print(add_array)
print(mul_array)
```

```python
arr5 = np.arange(12)
print(arr5)
```

```python
arr6=arr5.reshape(6,2) # reshape is used to decide the dimension of the array
print(arr6)
```

```python
arr6=arr5.reshape(3,4)
print(arr6)
```

```python
# statistics
print('Mean:',np.mean(arr5))
```

```python
print('Median:',np.median(arr5))
```

```python
print('Std dev:',np.std(arr5))
```

```python
# dot product 
a=np.array([[1,2,],[3,4]])
b=np.array([[5,6],[7,8]])
print('Dot Product:',np.dot(a,b))
```
