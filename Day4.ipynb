```python
# multi variate - heatmap - correlation
num_cols = df.select_dtypes(include='number')
num_cols
```

```python
cm=num_cols.corr()   # corr - correlation
cm
```

```python
# presenting correlation in a visual way is called as heatmap
sns.heatmap(cm, annot  = True)  # annot means annotation 
```

```python
sns.heatmap(cm, annot  = True, cmap='coolwarm')  # cmap means colormap
```
